```python
def array_reshape(arr, row: int = -1, col: int = -1):
    
    try:
        arr = np.array(arr)
        return arr.reshape((row, col))
    except (TypeError, ValueError) as e:
        raise ValueError(f"Invalid input: {e}")
   
```

```python
vec=[1,2,3,4,5,6]
array_reshape(vec, 2, 3)
```

```python
def  convert_dtype(arr, dtype):
    try:
        arr=np.asarray(arr)
        return arr.astype(dtype)
    
    except TypeError as e:
        raise TypeError(f"invalid target data type:{dtype}") from e
    
    except ValueError as e:
        raise ValueError(f"can not convert value to {dtype}") from e
```

```python
def check_dtype(arr, dtype):
  
    if not isinstance(arr, np.ndarray):
            raise TypeError  ("input must be a numpy array")
        
    dtype = np.dtype(dtype)
    return arr.dtype == dtype          
```

```python
def init_weights(n_feature, method = "zeros", seed = None):
    if not isinstance(n_feature, int):
        raise TypeError(f"{n_feature} must be an integer")
    
    if n_feature<=0:
        raise TypeError(f"{n_feature} must be greater than zero")
    if method=="zeros":
        return np.zeros(n_feature)
    if method=="random":
        if seed is not None:
            np.random.seed(seed)
        return np.random.random(n_feature)* 0.01
    
    raise ValueError(f"unsupported method type {method}")        
    
```

```python
init_weights(4,"random",42)
```
