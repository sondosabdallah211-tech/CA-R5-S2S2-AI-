```python
import numpy as np
# import pandas as pd
import matplotlib.pyplot as plt
# import seaborn as sns
# from sklearn.model_selection import train_test_split
# from sklearn.linear_model import LinearRegression
# from sklearn.metrics import  mean_squared_error, mean_absolute_error, r2_score, root_mean_squared_error
```

```python
x = np.array([50, 60, 70, 80, 90])
y = np.array([150, 180, 210, 240, 270])
```

```python
class LinearRegressionGD:
    
    def __init__(self, learning_rate=0.001, n_iters=100):
        self.learning_rate = learning_rate
        self.n_iters = n_iters
        self.theta_0 = 0
        self.theta_1 = 0
        
    def fit(self, x, y):
        self.cost_history = []
        
        for i in range(self.n_iters):
            y_pred = self.theta_0 + self.theta_1 * x
            error = y_pred - y
            
            d_theta_0 = np.mean(error)
            d_theta_1 = np.mean(error * x)
            
            self.theta_0 -= self.learning_rate * d_theta_0
            self.theta_1 -= self.learning_rate * d_theta_1
            
            y_pred = self.theta_0 + self.theta_1 * x
            
            cost = np.mean((y_pred - y) ** 2)
            
            self.cost_history.append(cost)
            
            print(i + 1, self.theta_0, self.theta_1, cost)
            
    def predict(self, x):
        return  self.theta_0 + self.theta_1 * x      
     
     
    def plot_training(self):
        plt.plot(range(1, self.n_iters+1), self.cost_history) 
        plt.xlabel("Iterations")
        plt.ylabel("Cost")
        plt.title("Cost over Iterations")
        plt.show()        
```

```python
model = LinearRegressionGD(
    learning_rate = 0.00001,
    n_iters = 100
)

model.fit(x, y)

# 0.001 caused divergence, so we used 0.00001 because the cost decreases
# gradually and the model converges toward a solution.
```

```python
prediction = model.predict(70)

print("Predicted price =", prediction)
```

```python
model.plot_training()
```

```python
y_pred = model.predict(x)

plt.scatter(x, y, label="Actual Data")
plt.plot(x, y_pred,'r-', label="Regression Line")

plt.xlabel("House Size (m²)")
plt.ylabel("House Price (thousands)")
plt.title("Linear Regression")
plt.legend()
plt.show()
```
