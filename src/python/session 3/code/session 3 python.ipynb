```python
def  train_model(model_name, *datasets, **Hyperpramemters):
    print(f"traning model: {model_name}")
    print("\n datasets")
    for dataset in datasets:
        print("_", dataset)
    print("\n Hyperpramemters")   
    for key, value in Hyperpramemters.items():
         print(f"{key}:{value}")
        
```

```python
train_model(
    "fraide detection",
    "train.csv",
    "valid.csv",
    "test.csv",
    learning_rate =0.01,
    batch_size =32,
    epochs=10,
    optimizer="adam",
    earlystopping=True
    
)
```
