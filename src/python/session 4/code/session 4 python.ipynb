```python
def factorial(num):
    '''
    Description of this func
    Arg:
    parameter num(int): this is int input for your num
    
    Returns:
    return int value 
    
    '''
    if num==0:
        return 1
    if num==1:
        return 1
    return num*factorial(num-1)

if __name__=="__main__":
    print(factorial(5))
        
```

```python
def isPrime(num):
    for i in range(2,num):
        if num %i==0:
            return False
    return True
isPrime(9)
```

```python
def reverse(word:str):
    '''
    Description of this func
    arg:
    parameter word(str):this is str input
    
    Returns:
    return str value
    '''
    lst=word.split(" ")
    print(lst)
    new_lst=lst[::-1]
    return" ".join(new_lst)
    
```

```python
'''
This file contains the code for generating a random password of a given length. 
The password is generated using a combination of charchters
'''
import random
def generate_password(length : int) -> str:
    '''  
    This function generates a random password of a given length.
    Args:
        length (int) : the length of the password to generate
    Returns:
        str : the generated password
    '''
    characters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789!@#$%^&*()_+-=~`"
    password = ""

    for _ in range(length):
        password += random.choice(characters)

    return password
```

```python
generate_password(12)
```
