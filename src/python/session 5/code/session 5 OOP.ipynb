```python
from abc import ABC, abstractmethod

class animal(ABC):
    
    @abstractmethod
    def move(self):
        pass

class bird(animal):
    def move(self):
        print("i can fly")
        
class cat(animal):
    def move(self):
        print("meow meow i can run")            
```

```python
a= bird()
a.move()
```

```python
class rectangle:
    length= 10
    width= 5
    def area(self):
        return self.length*self.width
    
r= rectangle()
print(r.length)
print(r.width)
print(r.area())
    
```

protected

```python
class rectangle:
    _length= 10
    _width= 5
    def area(self):
        return self._length*self._width
    
r= rectangle()
print(r._length)
print(r._width)
print(r.area())
```

private

```python
class rectangle:
    __length= 10
    __width= 5
    def area(self):
        return self.__length*self.__width
    
r= rectangle()
print(r._rectangle__length)
print(r._rectangle__width)
print(r.area())
```

```python
class parent:
    def __init__(self) -> None:
        self.__value= 10
        
class child(parent):        
    def __init__(self) -> None:
        super().__init__()
        self.__value= 20
```

```python
c= child()
print(c.__dict__)
```
