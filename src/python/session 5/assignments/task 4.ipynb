```python
class MultiplicationTable:
    """
    MultiplicationTable class handles the operation of generating
    a multiplication table for a given number.

    Attributes:
        number (int): The number used to generate the multiplication table.
    """
    def __init__(self, number: int):
        self.number = number
        
    def generate_table(self):
        """
        Generates and prints the multiplication table from 1 to 10.

        Returns:
            None: This method only prints the multiplication table.
        """
        
        for i in range(1, 11):
            result = self.number * i
            print(f"{self.number} x {i} = {result}")   
       
        
        
        
```

```python
number = int(input("Enter a number: "))
num=MultiplicationTable(number)
num.generate_table()
   
```

```python
class TwinPrimes:
    """
    Handles the operation of finding twin prime numbers.
    
    Attributes:
        limit (int): The upper limit for finding twin primes.
    """
    def __init__(self, limit: int):
        self.limit = limit
    
    def is_prime(self, num:int):
        """
        Checks if a number is prime.
        
        Args:
            num (int): The number to check for primality.
        
        Returns:
            bool: True if the number is prime, False otherwise.
        """
        for j in range(2, num):
                if num % j == 0:
                    return False
        return True
    
    def twin_primes(self):
        """
        Finds and prints twin prime pairs up to the specified limit.
        
        Returns:
            None: This method only prints the twin prime pairs.
        """
        for i in range(2, self.limit):
            if self.is_prime(i) and self.is_prime(i + 2):
                print(f"({i}, {i + 2})")
                        
```

```python
limit = int(input("Enter the limit for twin primes: "))
prime = TwinPrimes(limit)
prime.twin_primes()
```

```python
class BinaryTransform:
    """
    Handles the operation of converting a decimal number to binary.

    Attributes:
        num (int): The decimal number to be converted.
    """
    def __init__(self, num: int):
        self.num = num
        
    def decTobin(self):
        '''
        Converts the decimal number to binary representation.
        Returns:
            str: The binary representation of the decimal number.
        '''
        
        result = []
        while self.num > 0:
            remainder = self.num % 2
            result.append(remainder)
            self.num = self.num // 2
            
        result.reverse()
        return "".join(map(str, result))
        
```

```python
num = int(input("Enter a decimal number to convert to binary: "))
b=BinaryTransform(num)
b.decTobin()
```

```python
class perfectnumber:
    """
    Handles the operation of checking if a number is a perfect number.

    Attributes:
        num (int): The number to be checked.
    """
    def __init__(self, limit: int):
        self.limit = limit

    def check_perfect(self):
        '''
        Checks if the number is a perfect number.
        Returns:
            bool: True if the number is perfect, False otherwise.
            
        '''
        for i in range(0, self.limit):
            sum = 0
            for j in range(1, i):
                if i % j == 0:
                    sum += j
            if sum == i:
                print(i)        
                    
   
        
```

```python
limit = int(input("Enter the limit for perfect numbers: "))
p=perfectnumber(limit)
p.check_perfect()
```
